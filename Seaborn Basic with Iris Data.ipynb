```python
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
iris = sns.load_dataset("iris")
```

```python
iris.head()
```

### Distribution plots 

Histplot  
Join plot  
Pair plot  


```python
# Hist plot 

sns.histplot(data=iris, x = "sepal_length", kde = True, bins = 20) 

plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Count')
plt.show()
```

```python
# Join plot 

sns.jointplot(data=iris, x='sepal_length', y='sepal_width', kind='scatter', hue='species')
```

```python
# kind = kde 
sns.jointplot(data=iris, x='sepal_length', y='sepal_width', kind="kde", hue='species')
```

```python
# Pairplot
sns.pairplot(data=iris, hue='species', diag_kind='kde')
```

```python
# or 'hist' for histograms on the diagonal
sns.pairplot(data=iris, hue='species', diag_kind='hist')
```

## Correlation PLot 

```python
# Compute correlation matrix
corr = iris.drop('species', axis=1).corr()

# Plot correlation heatmap
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Iris Features')
plt.show()
```

```python
# Count Plot 

sns.countplot(data=iris, x='species')
plt.title('Count of Each Iris Species')
plt.show()
```

```python
#Box plot 

sns.boxplot(data=iris, x='species', y='sepal_width')
plt.title('Sepal Width Distribution by Species')
plt.show()
```

```python
#Violin plot 

sns.violinplot(data=iris, x='species', y='sepal_length')
plt.title('Sepal Length Distribution by Species (Violin Plot)')
plt.show()
```
